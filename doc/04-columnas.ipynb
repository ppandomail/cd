# Columnas

## Crear columnas

````python
import pandas as pd

# Dataset de ventas de una tienda escolar
datos = {
    'producto': [
        'Cuaderno', 'Lapicera', 'Mochila', 'Regla', 'Cartuchera',
        'Calculadora', 'Auriculares', 'Resaltadores', 'Guardapolvo', 'Compas',
        'Pendrive', 'Carpeta', 'Goma', 'Lapiz', 'Agenda'
    ],
    'categoria': [
        'Libreria', 'Libreria', 'Accesorios', 'Libreria', 'Accesorios',
        'Tecnologia', 'Tecnologia', 'Libreria', 'Indumentaria', 'Libreria',
        'Tecnologia', 'Libreria', 'Libreria', 'Libreria', 'Libreria'
    ],
    'precio': [
        1200, 350, 8500, 500, 3200,
        18500, 7600, 2100, 14500, 2600,
        9800, 950, 300, 250, 4300
    ],
    'cantidad_vendida': [
        15, 40, 5, 25, 8,
        3, 6, 12, 4, 10,
        7, 18, 30, 50, 9
    ],
    'sucursal': [
        'Centro', 'Centro', 'Norte', 'Sur', 'Norte',
        'Centro', 'Sur', 'Centro', 'Norte', 'Sur',
        'Centro', 'Norte', 'Sur', 'Centro', 'Sur'
    ],
    'forma_pago': [
        'Efectivo', 'Debito', 'Credito', 'Efectivo', 'Debito',
        'Credito', 'Credito', 'Debito', 'Credito', 'Efectivo',
        'Debito', 'Efectivo', 'Efectivo', 'Debito', 'Credito'
    ],
    'fecha': [
        '2024-03-01', '2024-03-01', '2024-03-02', '2024-03-02', '2024-03-03',
        '2024-03-03', '2024-03-04', '2024-03-04', '2024-03-05', '2024-03-05',
        '2024-03-06', '2024-03-06', '2024-03-07', '2024-03-07', '2024-03-08'
    ]
}

df = pd.DataFrame(datos)
df
````

* Se observa que en el `DataFrame` todavía no existe una columna que indique el importe total de cada venta
* Se necesita crear nuevas columnas para representar información que no aparece explícitamente en el dataset original, pero que puede obtenerse a partir de las columnas disponibles

````python
# A partir de las columnas precio y cantidad_vendida se creará una nueva variable: el importe total de cada venta
# importe_total = precio × cantidad_vendida
df['importe_total'] = df['precio'] * df['cantidad_vendida']
df
````

* Ahora el `DataFrame` tiene una nueva columna llamada `importe_total`
* Pandas realizó la multiplicación fila por fila
* Es decir, tomó el precio de cada producto y lo multiplicó por la cantidad vendida en esa misma fila

````python
# Verificar la nueva columna -> se observan solamente las columnas involucradas en el cálculo
# Esta vista permite comprobar si el cálculo tiene sentido
df[['producto', 'precio', 'cantidad_vendida', 'importe_total']]
````

````python
# Ordenar el DataFrame según la nueva columna 'importe_total' para ver cuáles fueron las ventas de mayor valor
df.sort_values('importe_total', ascending=False)[
    ['producto', 'categoria', 'precio', 'cantidad_vendida', 'importe_total']
]
````

````python
# Crear una columna con un valor constante
df['origen'] = 'Tienda escolar'
df
````

````python
# Crear una columna a partir de una condición
df['venta_alta'] = df['importe_total'] >= 50000

df[['producto', 'precio', 'cantidad_vendida', 'importe_total', 'venta_alta']]
````

````python
import numpy as np

# Crear una columna con etiquetas de texto
# Regla:
#   si la venta cumple la condición → escribir 'alta'
#   si no cumple la condición       → escribir 'normal'
df['tipo_venta'] = np.where(
    df['importe_total'] >= 50000,
    'alta',
    'normal'
)

df[['producto', 'importe_total', 'venta_alta', 'tipo_venta']]
````

````python
# Asignar valores usando loc, útil cuando se quiere asignar valores solo a las filas que cumplen cierto criterio
df['observacion'] = 'Sin observaciones'

# df.loc[condición, 'columna'] = nuevo_valor
df.loc[
    df['importe_total'] >= 50000,
    'observacion'
] = 'Revisar venta alta'

df[['producto', 'importe_total', 'tipo_venta', 'observacion']]
````

````python
# Revisar las columnas creadas
df.columns
````

````python
# Revisar si las nuevas columnas tienen sentido
df[
    [
        'producto',
        'precio',
        'cantidad_vendida',
        'importe_total',
        'tipo_venta',
        'observacion'
    ]
]
````

* La columna `importe_total` surge de multiplicar `precio` por `cantidad_vendida`
* La columna `venta_alta` indica con `True` o `False` si una venta supera cierto umbral
* La columna `tipo_venta` expresa una clasificación similar, pero con etiquetas de texto
* La columna `observacion` agrega una descripción para las ventas que se quieren destacar

* En análisis de datos, no alcanza con escribir una instrucción y seguir adelante
* Cada vez que se modifica el dataset, conviene verificar que el cambio haya producido el resultado esperado
* Otro punto importante es no crear demasiadas columnas sin necesidad. Agregar columnas puede enriquecer el análisis, pero también puede volver el `DataFrame` más difícil de leer si no se tiene claro para qué sirve cada nueva variable
* Una buena práctica es revisar el `DataFrame` después de cada transformación importante y conservar nombres de columnas descriptivos

## Modificar columnas

````python
import pandas as pd

# Dataset de trabajo con problemas intencionales
datos = {
    "producto": [
        "Cuaderno", "Lapicera", "Mochila", "Regla", "Cartuchera",
        "Calculadora", "Auriculares", "Resaltadores", "Guardapolvo", "Compas",
        "Pendrive", "Carpeta", "Goma", "Lapiz", "Agenda"
    ],
    "categoria": [
        "Libreria", "libreria", "Accesorios", " Libreria", "accesorios",
        "Tecnologia", "tecnologia", "Libreria ", "Indumentaria", "Libreria",
        "Tecnologia", " libreria", "Libreria", "libreria ", "Libreria"
    ],
    "precio": [
        1200, 350, 8500, 500, 3200,
        18500, 7600, 2100, 14500, 2600,
        9800, 950, 300, 250, 4300
    ],
    "cantidad_vendida": [
        15, 40, 5, 25, 8,
        3, 6, 12, 4, 10,
        7, 18, 30, 50, 9
    ],
    "sucursal": [
        "Centro", "Centro", "Norte", "Sur", "Norte",
        "Centro", "Sur", "Centro", "Norte", "Sur",
        "Centro", "Norte", "Sur", "Centro", "Sur"
    ],
    "forma_pago": [
        "Efectivo", "debito", "Credito", "Efectivo ", "Debito",
        "credito", "Credito", "Debito", "credito ", "Efectivo",
        "Debito", "efectivo", "Efectivo", "debito ", "Credito"
    ],
    "fecha": [
        "2024-03-01", "2024-03-01", "2024-03-02", "2024-03-02", "2024-03-03",
        "2024-03-03", "2024-03-04", "2024-03-04", "2024-03-05", "2024-03-05",
        "2024-03-06", "2024-03-06", "2024-03-07", "2024-03-07", "2024-03-08"
    ]
}

df = pd.DataFrame(datos)
df
````

* **Problemas**:
    * `"Libreria"`, `"libreria"`, `" Libreria"` y `"Libreria "`
    * `"Credito"`, `"credito"` o `"credito "`
* Este tipo de inconsistencias puede afectar filtros, conteos, agrupamientos y gráficos
* Antes de analizar datos categóricos o textuales, muchas veces se necesita revisar y modificar los valores de algunas columnas para corregir errores de escritura

````python
# La columna 'categoria' no está unificada
# Aparecen varias versiones de una misma categoria
df['categoria'].value_counts()
````

````python
# Revisar valores únicos antes de modificar
df['categoria'].unique()
````

````python
df['forma_pago'].unique()
````

````python
# Quitar espacios inncesarios
df['categoria'] = df['categoria'].str.strip()
df['categoria'].unique()
````

````python
df['forma_pago'] = df['forma_pago'].str.strip()
df['forma_pago'].unique()
````

````python
# Unificar mayúsculas y minúsculas
df['categoria'] = df['categoria'].str.lower()
df['categoria'].unique()
````

````python
df['forma_pago'] = df['forma_pago'].str.lower()
df['forma_pago'].unique()
````

````python
# Reemplazar valores con replace()
# Útil cuando se quiere corregir valores específicos
reemplazos_categoria = {
    'libreria': 'Libreria',
    'accesorios': 'Accesorios',
    'tecnologia': 'Tecnologia',
    'indumentaria': 'Indumentaria'
}

df['categoria'] = df['categoria'].replace(reemplazos_categoria)
df['categoria'].unique()
````

````python
reemplazos_pago = {
    'efectivo': 'Efectivo',
    'debito': 'Debito',
    'credito': 'Credito'
}

df['forma_pago'] = df['forma_pago'].replace(reemplazos_pago)
df['forma_pago'].unique()
````

````python
# Otra forma es crear una nueva columna, sin sobreescribir la original
# Se pueden encadenar transformaciones (aplicar varias transformaciones seguidas sobre una misma columna)
df = pd.DataFrame(datos)

df['categoria_limpia'] = (
    df['categoria']
    .str.strip()
    .str.lower()
    .replace({
        'libreria': 'Libreria',
        'accesorios': 'Accesorios',
        'tecnologia': 'Tecnologia',
        'indumentaria': 'Indumentaria'
    })
)

df[['categoria', 'categoria_limpia']]
````

````python
# Modificar columnas numéricas
# precios con un aumento del 10%.
df['precio'] = df['precio'] * 1.10
df[['producto', 'precio']].head()
````

En general, antes de modificar columnas conviene seguir esta rutina:

```text
revisar los valores originales
aplicar la transformación
verificar el resultado
```

Esta forma de trabajo reduce errores y ayuda a construir un análisis más confiable.

## Organizar columnas

* Organizar columnas significa revisar y ajustar la estructura del `DataFrame` para que sea más clara y fácil de usar
* Mirar la estructura del `DataFrame`:
    * cómo se llaman sus columnas
    * cuáles conservar
    * cuáles eliminar
    * en qué orden conviene presentarlas

````python
import pandas as pd

datos = {
    "Producto vendido": [
        "Cuaderno", "Lapicera", "Mochila", "Regla", "Cartuchera",
        "Calculadora", "Auriculares", "Resaltadores", "Guardapolvo", "Compas"
    ],
    "Cat": [
        "Libreria", "Libreria", "Accesorios", "Libreria", "Accesorios",
        "Tecnologia", "Tecnologia", "Libreria", "Indumentaria", "Libreria"
    ],
    "Precio Unitario": [
        1200, 350, 8500, 500, 3200,
        18500, 7600, 2100, 14500, 2600
    ],
    "Cant.": [
        15, 40, 5, 25, 8,
        3, 6, 12, 4, 10
    ],
    "Sucursal Venta": [
        "Centro", "Centro", "Norte", "Sur", "Norte",
        "Centro", "Sur", "Centro", "Norte", "Sur"
    ],
    "Pago": [
        "Efectivo", "Debito", "Credito", "Efectivo", "Debito",
        "Credito", "Credito", "Debito", "Credito", "Efectivo"
    ],
    "Fecha venta": [
        "2024-03-01", "2024-03-01", "2024-03-02", "2024-03-02", "2024-03-03",
        "2024-03-03", "2024-03-04", "2024-03-04", "2024-03-05", "2024-03-05"
    ],
    "Código interno": [
        "A001", "A002", "B010", "A003", "B011",
        "T020", "T021", "A004", "I030", "A005"
    ]
}

df = pd.DataFrame(datos)
df
````

* Algunas columnas tienen espacios, como `Producto vendido` o `Precio Unitario`
* Otras usan abreviaturas, como `Cat` o `Cant.`
* También hay nombres que podrían ser más descriptivos, como `Pago`, que en realidad representa la forma de pago
* Estos detalles pueden parecer menores, pero afectan la claridad del código

````python
# Revisar nombres actuales
list(df.columns)
````

```text
¿Los nombres son claros?
¿Hay abreviaturas poco descriptivas?
¿Hay espacios o signos que puedan incomodar al escribir código?
¿Los nombres siguen un mismo estilo?
¿Hay columnas que no vamos a necesitar?
¿Conviene reordenar las columnas para que la tabla se lea mejor?
```

Estas preguntas nos ayudan a tomar decisiones antes de modificar la estructura del `DataFrame`

````python
# Renombrar columnas con rename
# Para que tengan nombres más claros y consistentes
df = df.rename(columns={
    'Producto vendido': 'producto',
    'Cat': 'categoria',
    'Precio Unitario': 'precio_unitario',
    'Cant.': 'cantidad_vendida',
    'Sucursal Venta': 'sucursal',
    'Pago': 'forma_pago',
    'Fecha venta': 'fecha'
})
df
````

````python
list(df.columns)
````

````python
# Eliminar columnas con drop()
# Porque se decide que no era necesaria para este análisis
df = df.drop(columns=['Código interno'])
df
````

Antes de eliminar una columna conviene preguntarse:

```text
¿Esta columna aporta información útil para el análisis?
¿La información aparece repetida en otra columna?
¿Podría necesitarla más adelante para verificar o cruzar datos?
¿Estoy trabajando sobre una copia o sobre el único DataFrame disponible?
```

En general, si no se está seguro, puede ser mejor conservar la columna o crear una versión organizada del `DataFrame` en una nueva variable

````python
# Reordenar columnas
# Para que la tabla quede presentada de una manera más cómoda
columnas_ordenadas = [
    "producto",
    "categoria",
    "precio_unitario",
    "cantidad_vendida",
    "sucursal",
    "forma_pago",
    "fecha"
]
df = df[columnas_ordenadas]
df
````
