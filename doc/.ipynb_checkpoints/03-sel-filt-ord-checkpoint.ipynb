# Selección, filtros y ordenamientos

````python
import pandas as pd

# Dataset de ventas de una libreria
datos = {
    'producto': [
        'Cuaderno', 'Lapicera', 'Mochila', 'Regla', 'Cartuchera',
        'Cuaderno', 'Marcadores', 'Carpeta', 'Goma', 'Lapiz'
    ],
    'categoria': [
        'Libreria', 'Libreria', 'Accesorios', 'Libreria', 'Accesorios',
        'Libreria', 'Libreria', 'Libreria', 'Libreria', 'Libreria'
    ],
    'precio': [
        1200, 350, 8500, 500, 3200,
        1250, 1800, 950, 300, 250
    ],
    'cantidad_vendida': [
        15, 40, 5, 25, 8,
        12, 10, 18, 30, 50
    ],
    'sucursal': [
        'Centro', 'Centro', 'Norte', 'Sur', 'Norte',
        'Sur', 'Centro', 'Norte', 'Sur', 'Centro'
    ],
    'fecha': [
        '2024-03-01', '2024-03-01', '2024-03-02', '2024-03-02', '2024-03-03',
        '2024-03-03', '2024-03-04', '2024-03-04', '2024-03-05', '2024-03-05'
    ]
}

df = pd.DataFrame(datos)
df
````

## Selección de columnas en un dataframe

* Un DataFrame puede tener muchas columnas
* Algunas pueden ser necesarias para una pregunta determinada y otras no
* Por ejemplo, si se quiere comparar productos y precios, se puede seleccionar dos columnas: **producto** y **precio**

````python
# Seleccionar la columna precio como Serie (estructura de una sola dimensión)
serie_precio = df['precio']
serie_precio
````

````python
# Seleccionar la columna precio como DataFrame con una sola columna
df_precio = df[['precio']]
df_precio
````

````python
# Seleccionar las columnas producto y precio como DataFrame formado por las columnas seleccionadas
df_prod_precio = df[['producto', 'precio']]
df_prod_precio
````

````python
# Aplicar una operación sobre una columna (min(), max(), mean(), sum(), ...)
df['precio'].mean()
````

````python
df['precio'].sum()
````

Selección de columnas según la pregunta que se quiere responder, permite enfocar el análisis en las variables que se necesitan
1. Formular una pregunta de análisis
2. Identificar que columnas se necesitan
3. Aplicar las operaciones correspondientes (en Pandas)
4. Interpretar resultados

## Selección de filas en un dataframe

* Las columnas representan variables
* Las filas representan observaciones, registros o casos
* Seleccionar filas significa elegir qué registros del dataset se quiere observar

````python
# Seleccionar una fila con iloc mediante posiciones numéricas como Serie
df.iloc[0]
````

````python
# Seleccionar una fila con iloc mediante posiciones numéricas como DataFrame
df.iloc[[0]]
````

````python
# Seleccionar varias filas consecutivas como DataFrame
df.iloc[0:3]
````

````python
# Seleccionar varias filas NO consecutivas
df.iloc[[0, 3, 9]]
````

````python
# Seleccionar filas y columnas con iloc
# [fila, columna]
df.iloc[0, 0]
````

````python
df.iloc[[0, 1], [1, 2, 3]]
````

* Con iloc, se usan posiciones numéricas
* Para evitar confusiones, puede ser útil revisar primero:
    * el tamaño del DataFrame con shape y
    * el orden de las columnas con list(df.columns)

## Selección con etiquetas usando loc

* ```iloc``` selecciona por posición
* ```loc``` selecciona por etiqueta
* Se puede indicar nombres de columnas y etiquetas del índice

````python
# Seleccionar la fila que está en la primera posición
df.iloc[0]
````

````python
# Selecciona la fila cuya etiqueta de índice es 0
df.loc[0]
````

* En este dataset, ambas instrucciones devuelven la misma fila porque el índice coincide con las posiciones
* Pero esto no siempre será así
* Si el índice estuviera formado por códigos, fechas o nombres, loc usaría esas etiquetas, no la posición física de la fila

````python
# Seleccionar varias filas con loc
df.loc[[0, 3, 5]]
````

````python
# Seleccionar una columna con loc
# [filas, columnas]
df.loc[:, 'precio']
````

````python
# Seleccionar varias columnas con loc
df.loc[:, ['producto', 'precio']]
````

````python
df.loc[0:3, ['producto', 'precio']]
````

## Filtros con condiciones

* Filtrar significa quedarse solamente con los registros que cumplen cierto criterio
* Por ejemplo, se podría querer ver solo ...
    * las ventas realizadas en la sucursal "Centro"
    * los productos de la categoría "Libreria"
    * las ventas con precio mayor a 1000
    * los registros cuya cantidad vendida supere cierto valor
* Esta forma de selección es fundamental porque permite responder preguntas más específicas sobre los datos
* Ya no se elijen filas por su posición ni por su etiqueta, sino por lo que esas filas contienen

````python
# Construir una condición
# Devuelve una máscara booleana: serie que indica qué filas cumplen la condición
df['precio'] > 1000
````

````python
# Aplicar la máscara booleana para filtrar el DataFrame
df[df['precio'] > 1000]
````

````python
# Aplicar función lambda
df[df['precio'].apply(lambda x: x > 1000)]
````

````python
# Filtrar con valores categóricos
df[df['sucursal'] == 'Centro']
````

````python
# Aplicar función lambda
df[df['sucursal'].apply(lambda x: x == 'Centro')]
````

````python
# Filtrar valores incluidos en una lista
df[df['sucursal'].isin(['Centro', 'Norte'])]
````

````python
# Filtrar textos que contienen una palabra
df[df['producto'].str.contains('Cuad')]
````

````python
# Filtrar texto mediante expresiones regulares (patrón)
df[df['fecha'].str.match('2024-..-01')]
````

````python
# Combinar filtros
# df[(condición_1) & (condición_2)]
# la categoría es Libreria
# el precio es mayor que 500
df[(df['categoria'] == 'Libreria') & (df['precio'] > 500)]
````

| Operador | Significado                           | Ejemplo                            |
|----------|---------------------------------------|------------------------------------|
| **&**    | deben cumplirse todas las condiciones | df[(condición_1) & (condición_2)]  |
| **\|**   | debe cumplirse al menos una condición | df[(condición_1) \| (condición_2)] |
| **~**    | niega una condición                   | df[~(condición_1)]                 |

## Ordenar datos en un dataframe

* Ordenar un ```DataFrame``` significa reorganizar sus filas según los valores de una o más columnas
* Esto permite leer mejor la información, encontrar rápidamente valores altos o bajos, comparar registros y construir primeros rankings

````python
# Ordenar un dataframe según los valores de una columna
# Por default ordena de menor a mayor
df.sort_values('precio')
````

````python
df.sort_values('precio', ascending=False)
````

````python
# Ordenar columnas de texto (en forma alfabetica)
df.sort_values('producto')
````

````python
df.sort_values('producto', ascending=False)
````

````python
# Ordenar varias columnas
df.sort_values(['categoria', 'precio'])
````

````python
# Ordenar varias columnas por distintos criterios
df.sort_values(['categoria', 'precio'], ascending=[True, False])
````

````python
# Combinar filtros, ordenamiento y selección de columnas
# SQL:
#  SELECT producto, categoria, precio
#  FROM df
#  WHERE categoria == 'Libreria'
#  ORDER BY precio desc
df[df['categoria'] == 'Libreria'].sort_values('precio', ascending=False)[['producto', 'categoria', 'precio']]
````

## Valores extremos y primeros rankings

````python
# Precios altos
df.sort_values('precio', ascending=False).head(3)
````

````python
# Precios bajos
df.sort_values('precio', ascending=False).tail(3)
````

````python
# Obtener los mayores valores con nlargest()
df.nlargest(3, 'precio')
````

````python
# Obtener los menores valores con nsmallest()
df.nsmallest(3, 'precio')
````
