* Las cuentas registradas durante la cena (`Dinner`) tienen un promedio de **20.80**, mientras que las cuentas registradas durante el almuerzo (`Lunch`) tienen un promedio de **17.17**
* Esto sugiere que, en estos datos, el consumo promedio es mayor durante la cena

````python
# ¿la propina promedio también cambia entre almuerzo y cena?
promedio_propina_almuerzo = df[df['time'] == 'Lunch']['tip'].mean()
promedio_propina_cena = df[df['time'] == 'Dinner']['tip'].mean()

comparacion_propinas = pd.DataFrame({
    'horario': ['Lunch', 'Dinner'],
    'promedio_propina': [promedio_propina_almuerzo, promedio_propina_cena]
})

comparacion_propinas['promedio_propina'] = comparacion_propinas['promedio_propina'].round(2)
comparacion_propinas
````

* La propina promedio también es mayor durante la cena
* En el almuerzo (`Lunch`), la propina promedio es de **2.73**
* En la cena (`Dinner`), la propina promedio es de **3.10**
* Este resultado acompaña lo que se vió con el total de la cuenta: en este dataset, tanto el promedio de cuenta como el promedio de propina son más altos durante la cena

* En este caso, ambos valores están muy cerca, por lo que resumen la columna de manera bastante similar

# Agrupamiento

## Agrupar datos con ```groupby()```

Separar un dataset en partes según los valores de una columna y luego calcular algún resumen dentro de cada parte

````python
# ¿Cuál es el promedio de cuenta para cada dia?
# observed=True  le indica a Pandas que trabaje con las categorias observdas en los datos
promedio_cuenta_por_dia = df.groupby('day', observed=True)['total_bill'].mean()
promedio_cuenta_por_dia = promedio_cuenta_por_dia.round(2)
promedio_cuenta_por_dia
````

````python
# ¿Cuál es el total facturado por día?
total_facturado_por_dia = df.groupby('day', observed=True)['total_bill'].sum()
total_facturado_por_dia = total_facturado_por_dia.round(2)
total_facturado_por_dia
````

````python
# Calcular la cantidad de registros por grupo
cantidad_cuentas_por_dia = df.groupby('day', observed=True)['total_bill'].count()
cantidad_cuentas_por_dia
````

````python
# Combinar varios resultados por grupo
resumen_por_dia = pd.DataFrame({
    'cantidad_cuentas': cantidad_cuentas_por_dia,
    'total_facturado': total_facturado_por_dia,
    'promedio_cuenta': promedio_cuenta_por_dia
})
resumen_por_dia
````

````python
# Calcular varios indicadores con agg()
# agg viene de 'aggregate': aplicar 1 ó + funciones de resumen sobre cada grupo
resumen_por_dia_agg = df.groupby('day', observed=True)['total_bill'].agg(
    cantidad_cuentas='count',
    total_facturado='sum',
    promedio_cuenta='mean'
)
resumen_por_dia_agg = resumen_por_dia_agg.round(2)
resumen_por_dia_agg
````

````python
# Convertir el índice en una columna
# para que 'day' vuelva a aparecer como una columna común del DataFrame
resumen_por_dia_agg = resumen_por_dia_agg.reset_index()
resumen_por_dia_agg
````

````python
# Agrupar por más de una columna
# ¿Cuántas cuentas hay para cada combinación de día y horario?
cuentas_por_dia_y_horario = df.groupby(['day', 'time'], observed=True)['total_bill'].count()
cuentas_por_dia_y_horario
````

````python
cuentas_por_dia_y_horario = cuentas_por_dia_y_horario.reset_index()

cuentas_por_dia_y_horario = cuentas_por_dia_y_horario.rename(
    columns={'total_bill': 'cantidad_cuentas'}
)

cuentas_por_dia_y_horario
````

````python
# ¿Cuál es el promedio de cuenta para cada combinación de día y horario?
promedio_cuenta_por_dia_y_horario = df.groupby(['day', 'time'], observed=True)['total_bill'].mean()
promedio_cuenta_por_dia_y_horario = promedio_cuenta_por_dia_y_horario.round(2).reset_index()
promedio_cuenta_por_dia_y_horario = promedio_cuenta_por_dia_y_horario.rename(
    columns={'total_bill': 'promedio_cuenta'}
)
promedio_cuenta_por_dia_y_horario
````

````python
# Calcular varios indicadores por combinación de grupos
resumen_dia_horario = df.groupby(
    ['day', 'time'],
    observed=True
).agg(
    cantidad_cuentas=('total_bill', 'count'),
    total_facturado=('total_bill', 'sum'),
    promedio_cuenta=('total_bill', 'mean'),
    promedio_propina=('tip', 'mean')
)

resumen_dia_horario = resumen_dia_horario.round(2).reset_index()
resumen_dia_horario.sort_values(
    by='total_facturado',
    ascending=False
)
````

````python
# Filtrar grupos con pocos registros
# Para evitar interpretaciones débiles
resumen_grupos_relevantes = resumen_dia_horario[
    resumen_dia_horario['cantidad_cuentas'] >= 10
]

resumen_grupos_relevantes
````

## Valores extremos y outliers

* Cuando se ordenan los datos, los valores extremos aparecen con facilidad
* Un valor extremo es un dato que está muy por encima o muy por debajo de la mayoría de los valores observados
* Un outlier puede ser:
    * un error de carga o medición;
    * un caso real poco frecuente;
    * una situación especial que merece análisis;
    * un dato válido, pero muy diferente del resto.

````python
# Observar un valor extremo con más contexto
# Revisar el registro con la cuenta más alta del dataset
df.loc[df['total_bill'].idxmax()]
````

* Se observa que corresponde a una mesa de **3 personas**, durante ```Dinner```, un sábado (```Sat```). También registra una propina de **10.00**
* Esta información ayuda a contextualizar el valor
* No se observa ningún elemento que indique automáticamente que este dato sea un error

````python
# Observar el valor mínimo
df.loc[df['total_bill'].idxmin()]
````

*  Se observa que corresponde a una mesa de **1 persona**, durante ```Dinner```, un sábado (```Sat```). La propina registrada es de **1.00**
*  Una cuenta pequeña puede ser razonable si corresponde a una sola persona

Detectar un valor extremo no significa modificarlo automáticamente. Antes de decidir qué hacer con un outlier, conviene hacerse algunas preguntas:
* ¿el valor es posible dentro del contexto?
* ¿hay otras columnas que ayuden a explicarlo?
* ¿podría ser un error de carga o medición?
* ¿afecta mucho los promedios u otros indicadores?
* ¿el objetivo del análisis requiere conservarlo, marcarlo o tratarlo de alguna manera?

En algunos casos, un outlier puede ser un error y habrá que corregirlo o excluirlo. En otros casos, puede ser un dato real y valioso para el análisis. Por eso, en esta etapa no vamos a eliminar estos valores. Los vamos a interpretar como señales para observar con más atención

# Tablas dinámicas

* Permite cruzar variables
* Por ejemplo, se puede colocar los días en las filas, los horarios en las columnas y dentro de la tabla mostrar un indicador, como el promedio de cuenta o la cantidad de registros
* Esto no reemplaza a groupby(). Es otra forma de resumir datos, especialmente útil cuando se quiere comparar categorías en forma de tabla

````python
# ¿Cuál es el promedio de cuenta según el día y el horario?
# Agrupación con groupby
promedio_groupby = df.groupby(
    ['day', 'time'],
    observed=True
)['total_bill'].mean().round(2)

promedio_groupby
````

````python
# Uso de tabla dinámica con pivot_table()
tabla_promedio_cuenta = df.pivot_table(
    values='total_bill',        # indica qué columna numérica se quiere resumir
    index='day',                # indica qué variable se ubicará en las filas
    columns='time',             # indica qué variable se ubicará en las columnas
    aggfunc='mean',             # indica qué cálculo se aplicará sobre los datos
    observed=True
).round(2)

tabla_promedio_cuenta
````

* La tabla dinámica muestra la misma información que se obtuvo con groupby(), pero organizada de otra manera
* Ahora cada fila representa un día y cada columna representa un horario

````python
# ¿Cuántas cuentas hay para cada combinación de día y horario?

tabla_cantidad = df.pivot_table(
    values='total_bill',
    index='day',
    columns='time',
    aggfunc='count',
    observed=True
)

tabla_cantidad
````

````python
# Reemplazar valores vacíos en una tabla de conteos
tabla_cantidad = df.pivot_table(
    values='total_bill',
    index='day',
    columns='time',
    aggfunc='count',
    observed=True,
    fill_value=0              # reemplaza valores vacíos
)

tabla_cantidad
````

````python
# Agregar totales a una tabla dinámica
tabla_total_con_totales = df.pivot_table(
    values='total_bill',
    index='day',
    columns='time',
    aggfunc='sum',
    observed=True,
    margins=True            # agrega una fila y una columna adicional con resúmenes generales
).round(2)

tabla_total_con_totales
````

````python
# Tablas dinámicas con más de un valor
tabla_dos_valores = df.pivot_table(
    values=['total_bill', 'tip'],
    index='day',
    columns='time',
    aggfunc='mean',
    observed=True
).round(2)

tabla_dos_valores
````

````python
# Tablas dinámicas con más de una función
tabla_varias_funciones = df.pivot_table(
    values='total_bill',
    index='day',
    columns='time',
    aggfunc=['count', 'mean', 'sum'],
    observed=True
).round(2)

tabla_varias_funciones
````

## groupby() y pivot_table()

* Tanto ```groupby()``` como ```pivot_table()``` permiten resumir datos por grupos
* La diferencia principal está en la forma de organizar el resultado

* Con ```groupby()```, se suele obtener una tabla larga o una serie agrupada. Esa forma es muy útil para calcular, ordenar, filtrar y seguir trabajando con los datos
* Con ```pivot_table()```, el resultado se organiza como una tabla cruzada: una variable queda en las filas, otra en las columnas y los valores resumidos quedan dentro de la tabla
* No son herramientas rivales. Son dos formas distintas de resumir información. En general:
    * ```groupby()``` resulta cómodo para calcular resúmenes por grupos
    + ```pivot_table()``` resulta cómodo para comparar categorías cruzadas en forma de tabla
