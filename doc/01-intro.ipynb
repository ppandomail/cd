## Machine Learning (ML)

* Permite a las computadoras aprender y tomar decisiones sin necesidad de programación explícita
* Implica introducir datos en algoritmos para identificar patrones y realizar predicciones a partir de nuevos datos
* Se utiliza en diversas aplicaciones, como el reconocimiento de imagenes (distinguir entre perros y gatos), y voz, el procesamiento del lenguaje natural y los sistemas de recomendación

* En 2015 el universo digital estaba compuesto por 6 ZB de datos

    * 1 Hexabyte = 1000 Petabyte
    * 1 Petabyte = 1000 Terabyte
* 6 ZB en discos de 10TB $\rightarrow$ 644.245.094 discos

## Las 5 "V" de Big Data

<div align='center'><img src="images/5V.png" width="600"/></div>

* **Volumen**: cantidades que exigen un almacenamiento mucho más complejo y con una capacidad de almacenamiento enorme a nivel terabyte
* **Variedad**: los datos provienen de diferentes fuentes (datos estructurados y no estructurados)
* **Velocidad**: momento oportuno para que no pierdan valor
* **Valor**: información que puedan tener los datos para alcanzar el objetivo que se está buscando
* **Veracidad**: calidad en cuanto al resguardo de los datos. Problemas:
    * registros incompletos o erróneos
    * datos faltantes
    * información que, proveniente de diferentes fuentes, son discrepantes

## Ejemplos en el uso de Big Data

* **Netflix** utiliza Big Data para sus sistemas de recomendación y algunas estrategias/sugerencias comerciales y de marketing basadas en el análisis FODA y PESTLE

* **Amazon** utiliza Big Data para predecir el comportamiento de los usuarios y mejorar su experiencia en la plataforma

* **Otros**:
    * Segmentación de clientes (hipermercados, aseguradoras, etc.)
    * Campañas electorales
    * Optimización de procesos de negocio (cadena de puntos de venta, correo, etc.)
    * Optimización de rendimiento personal (google fit, apple swatch)
    * Codificación de material genético (laboratorios, farmacias, hospitales, etc.)
    * Fraudes, cyber-ataques, perfil criminal
    * Tráfico, optimización de suministro (electricidad)
