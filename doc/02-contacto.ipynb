    * seleccionar información relevante

## ¿Por qué aprender a trabajar con datos?

* Vivimos rodeados de datos
* Tener datos no significa automáticamente comprender una situación
* Para que esos datos sean útiles, necesitamos hacerles preguntas, revisarlos, ordenarlos, limpiarlos, analizarlos e interpretarlos
* Trabajar con datos implica combinar herramientas técnicas con interpretación

## Dato, información y conocimiento

$$\text{datos} \rightarrow \text{información} \rightarrow \text{conocimiento}$$

* **Dato**: es un valor registrado. Ejemplo: 8
* **Información**: dato aparece dentro de un contexto. Ejemplo: la nota de Pepe en matemáticas fue 8
* **Conocimiento**: aparece cuando podemos interpretar esa información, relacionarla con otros datos y utilizarla para tomar decisiones o comprender mejor una situación. Ejemplo: Pepe obtuvo un 8 en Matemática. Su promedio anterior era 6,5 y en las últimas semanas aumentó la cantidad de horas de estudio. Su desempeño parece haber mejorado

## ¿Qué significa analizar datos?

* Observar un conjunto de datos con el objetivo de comprender mejor una situación, responder preguntas o tomar decisiones fundamentadas
* No se trata solamente de aplicar fórmulas o ejecutar código
* El análisis de datos suele ser un proceso que incluye varias etapas:
    1. comprender el contexto
    2. observar los datos
    3. comenzar a hacer preguntas: ¿Cuál es el promedio? ¿Qué diferencias aparecen entre distintas categorías?
    4. interpretar los resultados
    5. comunicar lo encontrado mediante tablas, gráficos, dashboards, informes o explicaciones escritas

## ¿Qué es un dataset?

* Es un conjunto organizado de datos
* En muchos casos, se presenta como una tabla formada por filas y columnas
* Cada fila suele representar una observación, un registro o un caso
* Cada columna representa una característica, atributo o variable que se registró para cada observación
* Ejemplo:

|nombre	| edad	| curso	| horas_estudio	| nota |
| --    | --    | --    | --            | --   |
| Ana	| 15	| 3A	| 4	            | 8    |
| Bruno	| 16	| 3B	| 2	            | 6    |
| Carla	| 15	| 3A	| 6	            | 9    |
| Diego	| 17	| 4A	| 1	            | 5    |
| Elena	| 16	| 3B	| 5	            | 8    |

* Esta forma de organizar los datos es muy frecuente porque permite comparar casos, resumir información, buscar patrones y hacer preguntas como: ¿Los estudiantes que estudiaron más horas obtuvieron mejores notas?

## Filas, columnas, variables, observaciones y valores

* Cuando observamos un dataset tabular, es importante reconocer sus partes principales:

| Elemento	| Qué representa                      | Ejemplo                 |
| --------- |-------------------------------------|-------------------------|
| Fila	    | Una observación, registro o caso    | Un estudiante           |
| Columna	| Una variable o característica       | La nota                 |
| Valor	    | Un dato concreto dentro de la tabla | 8                       |
| Dataset	| El conjunto completo de datos       | La tabla de estudiantes |

## Tipos de variables

No todas las columnas de un dataset representan el mismo tipo de información. Algunas contienen números, otras contienen categorías, otras fechas y otras textos.

| Tipo de variable | Que representa                                                 | Ejemplos                          |
|------------------|----------------------------------------------------------------|-----------------------------------|
| numérica         | cantidad o medidas                                             | edad, notas, precio               |
| num. discreta    | valores contables                                              | cant. de estudiantes              |
| num. continua    | valores dentro de un rango                                     | altura, peso, temperatura         |
| categóricas      | grupos, clases o etiquetas                                     | curso, pcia, tipo prod, color     |
| ordinales        | categorías que tienen un orden natural                         | bajo/medio/alto                   |
| temporales       | fechas, horas o momentos en el tiempo                          | fecha de compra, fecha nacimiento |
| texto            | frases, comentarios, descripciones u otros fragmentos escritos | comentario, desc prod, observac   |

En general, antes de analizar una columna conviene preguntarnos: ¿qué tipo de información contiene? Esa respuesta nos ayuda a decidir qué operaciones tienen sentido y cuáles no.

## Primer dataset en Python

* Biblioteca Pandas
* Pandas permite trabajar con datos organizados en forma de tabla (DataFrame)
* Un DataFrame es una tabla de datos: tiene filas, columnas y valores
* Ejemplo: dataset de estudiantes
* diccionario de Python → DataFrame de Pandas

````python
# Importamos la biblioteca Pandas
# Por convención, se suele importar con el alias pd

import pandas as pd

# Creamos un diccionario de Python
# Cada clave del diccionario será el nombre de una columna
# Cada lista contiene los valores de esa columna

datos = {
    "nombre": ["Ana", "Bruno", "Carla", "Diego", "Elena"],
    "edad": [15, 16, 15, 17, 16],
    "curso": ["3A", "3B", "3A", "4A", "3B"],
    "horas_estudio": [4, 2, 6, 1, 5],
    "nota": [8, 6, 9, 5, 8]
}

# Convertimos el diccionario en un DataFrame de Pandas
# A partir de este momento, los datos quedan organizados como una tabla

df = pd.DataFrame(datos)

# Mostramos el DataFrame

df
````

* Se observa:
    * las columnas describen variables
    * las filas representan observaciones
    * el índice identifica las filas dentro del DataFrame

## Hacer preguntas sobre un dataset

* Simples y directas: ¿cuántos estudiantes hay en el dataset?
* Relacionan 2 o más variables: ¿los estudiantes que estudiaron más horas obtuvieron mejores notas?
* Complejas: ¿estudiar más horas causa mejores resultados?

## Primeras respuestas usando pandas

````python
cantidad_estudiantes = len(df)
nota_mas_alta = df["nota"].max()
nota_mas_baja = df["nota"].min()
promedio_notas = df["nota"].mean()
estudiantes_por_curso = df["curso"].value_counts()
````

* Una salida como: 7.2 no dice demasiado por sí sola. Pero si sabemos que representa el promedio de notas de un grupo de estudiantes, entonces se convierte en información útil
* En análisis de datos, muchas veces vamos a realizar este recorrido:

$$\text{código} \rightarrow \text{resultado} \rightarrow \text{interpretación}$$

* El código produce un resultado. Luego debemos leer ese resultado dentro del contexto del problema
* Por eso, cada vez que obtengamos una tabla, un número o un gráfico, conviene hacernos algunas preguntas:
    * ¿Qué representa este resultado?
    * ¿A qué columna o variable se refiere?
    * ¿En qué contexto debe interpretarse?
    * ¿Qué conclusión puedo sacar?
    * ¿Qué conclusión no debería sacar todavía?
* Esta costumbre ayuda a evitar interpretaciones apresuradas o conclusiones que los datos no sostienen

## Los límites de un dataset

* Un dataset puede ayudarnos a comprender una situación, pero no contiene toda la realidad.
* Hay muchas otras preguntas que no podemos responder con este dataset
    * ¿Qué tema se evaluó?
    * ¿La evaluación fue fácil o difícil?
* Un dataset siempre es una representación parcial de una situación
* Antes de sacar conclusiones, conviene preguntarnos:
    * ¿Qué información tenemos?
    * ¿Qué información falta?
    * ¿Cómo fueron obtenidos estos datos?
    * ¿Qué tan confiables son?
    * ¿Qué conclusiones permite sostener este dataset?
    * ¿Qué conclusiones serían exageradas?
* Esta mirada crítica es parte esencial del trabajo con datos

## Los datasets reales suelen tener problemas

* En situaciones reales, los datasets suelen ser más desordenados en comparación al dataset de estudiantes
* Podemos encontrar problemas como estos:

| Problema                     | Ejemplo                                                       |
|------------------------------|---------------------------------------------------------------|
| Datos faltantes              | Una fila no tiene cargada la edad o la nota                   |
| Valores duplicados           | El mismo estudiante aparece dos veces por error               |
| Nombres de columnas confusos | Una columna se llama col1 o dato_final                        |
| Espacios innecesarios        | El valor " 3A " aparece con espacios antes o después          |
| Diferencias de escritura     | Un curso aparece como 3A, 3a, 3 A o Tercero A                 |
| Tipos de datos incorrectos   | Una nota aparece como texto en lugar de número                |
| Valores imposibles           | Una edad de -5 o una nota de 25 si la escala va de 1 a 10     |
| Fechas mal cargadas          | Una misma fecha aparece como 2024-05-10, 10/05/2024 o mayo 10 |

* Estos problemas no son detalles menores. Si no los detectamos, pueden afectar los resultados del análisis
* Por eso, una parte muy importante del trabajo con datos consiste en revisar la calidad del dataset antes de analizarlo
* Antes de confiar en un resultado, debemos revisar los datos que usamos para obtenerlo.

## Describir columnas no numérica

* Cuando usamos df.describe(), Pandas resume automáticamente las columnas numéricas
* Pero un dataset también puede tener columnas no numéricas, como producto, categoria o sucursal.
* Si queremos obtener un resumen de todas las columnas, podemos usar ```include='all'```

````python
df.describe(include='all')
````

* Para las columnas no numéricas, Pandas muestra otras medidas:

| Medida       | Significado                               |
|--------------|-------------------------------------------|
| ```count```  | Cantidad de valores no nulos              |
| ```unique``` | Cantidad de valores distintos             |
| ```top```    | Valor que aparece con mayor frecuencia    |
| ```freq```   | Frecuencia del valor más repetido         |

* En columnas como producto, que pueden tener muchos valores distintos, a veces conviene usar otras herramientas, como ```value_counts()```, para ver con más detalle la frecuencia de cada valor.

````python
# ¿Cuántas ventas se registraron en cada sucursal?
df['sucursal'].value_counts()
````
